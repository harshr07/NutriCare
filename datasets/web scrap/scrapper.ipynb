```python

import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://ods.od.nih.gov/"

response = requests.get(url, timeout=20)
response.raise_for_status()

page = BeautifulSoup(response.text, "html.parser")
rows_data = []

for table in page.select("table"):
    for row in table.select("tr"):
        values = [cell.get_text(" ", strip=True) for cell in row.find_all(["th", "td"])]
        if any(values):
            rows_data.append(values)

nutrition_df = pd.DataFrame(rows_data)
nutrition_df.to_csv("raw_nutrition_data.csv", index=False)
nutrition_df.head()
```
